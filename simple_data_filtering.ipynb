```python
# DAY 8 - SIMPLE DATA FILTERING

students = [
    {"name": "Ananya", "age": 20, "marks": 85, "department": "AI & ML"},
    {"name": "Rahul", "age": 22, "marks": 68, "department": "Computer Science"},
    {"name": "Priya", "age": 21, "marks": 92, "department": "AI & ML"},
    {"name": "Kiran", "age": 19, "marks": 74, "department": "Electronics"},
    {"name": "Sneha", "age": 20, "marks": 78, "department": "Computer Science"},
    {"name": "Arjun", "age": 23, "marks": 88, "department": "AI & ML"}
]

print("STUDENT RECORDS")
print(students)

high_marks = [student for student in students if student["marks"] >= 75]

print()
print("STUDENTS WITH MARKS >= 75")
print(high_marks)

young_students = [student for student in students if student["age"] <= 21]

print()
print("STUDENTS WITH AGE <= 21")
print(young_students)

ai_students = [student for student in students if student["department"] == "AI & ML"]

print()
print("STUDENTS FROM AI & ML DEPARTMENT")
print(ai_students)

selected_students = [
    student
    for student in students
    if student["marks"] >= 75
    and student["age"] <= 21
    and student["department"] == "AI & ML"
]

print()
print("STUDENTS MATCHING ALL CONDITIONS")
print(selected_students)
```
